# Notebook 01: Read PS1 and understand the windows

This notebook is **only for exploring the data**. Run `main.py` to execute the
complete training-and-prediction pipeline.

We use **60 consecutive PS1 readings** for each example.
The data array has shape `(number_of_windows, 60)`.
Only when feeding the LSTM do we add a third dimension to get
`(number_of_windows, 60, 1)`.

The included `PS1_demo.csv` contains simulated example data, **not** real UCI PS1 data.

```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt

# Works if PyCharm runs notebooks from either the project folder or notebooks/.
PROJECT = Path.cwd()
if PROJECT.name == "notebooks":
    PROJECT = PROJECT.parent
sys.path.insert(0, str(PROJECT))

from data import load_windows

DATA_FILE = PROJECT / "data" / "PS1_demo.csv"
DATA_FORMAT = "cycles"  # change to "windows" for prepared 60-column Excel files
```

## 1. Load the data

```python
windows, cycle_ids = load_windows(DATA_FILE, data_format=DATA_FORMAT, windows_per_cycle=3)
print("PS1 windows:", windows.shape)
print("Cycle IDs:", cycle_ids.shape)
print("First window:", windows[0, :10])
```

## 2. Visualize a few healthy windows

```python
plt.figure(figsize=(10, 4))
for i in range(min(5, len(windows))):
    plt.plot(windows[i], label=f"Window {i}")
plt.title("Example PS1 windows")
plt.xlabel("Reading number inside the window")
plt.ylabel("Pressure value")
plt.legend()
plt.show()
```

## 3. Check the LSTM input shape

```python
lstm_input = windows[..., None]
print("Normal windows:", windows.shape)
print("LSTM input:", lstm_input.shape)
# There is only ONE feature here: the PS1 sensor reading.
```
